```python
# timestamp를 이용해서 period로 변환

dates = ["2019-01-01", "2020-03-01", "2021-06-01"]

ts_dates = pd.to_datetime(dates)
ts_dates
```

```python
pr_day = ts_dates.to_period(freq="D")
pr_day
```

```python
pr_month = ts_dates.to_period(freq="M")
pr_month
```

```python
pr_year = ts_dates.to_period(freq="Y")
pr_year
```

```python
# 시계열 데이터 만들기
ts_ms = pd.date_range(
    start="2024-01-01",
    end=None,
    periods=6,
    freq="MS",
    tz="Asia/Seoul"
)
ts_ms
```

```python
ts_me = pd.date_range(
    start="2024-01-01",
    periods=6,
    freq="ME",
    tz="Asia/Seoul"
)
ts_me
```

```python
ts_3m = pd.date_range(
    "2024-01-01",
    periods=6,
    freq="3MS",
    tz="Asia/Seoul"
)
ts_3m
```
